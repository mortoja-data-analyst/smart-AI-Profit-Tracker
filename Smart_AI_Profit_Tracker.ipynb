```python
# ===================================================
# PROJECT 2: SMART AI PROFIT & SALES TRACKER
# AUTHOR: MORTOJA SARDAR (DATA ANALYST)
# BUSINESS AUTOMATION AND ARTIFICIAL INTELLIGENCE
# ===================================================

import time
import IPython.display as ipd
from gtts import gTTS

print("=" * 55)
print("     🚀 SMART AI PROFIT & SALES TRACKER SYSTEM 🚀     ")
print("=" * 55 + "\n")

# 1. Automated Business Data Ingestion
total_sales = 8500.0   # Daily Gross Sales Revenue (USD)
total_expense = 5200.0 # Daily Total Operating Expenses (USD)

print("📊 [SYSTEM STATUS] Ingesting Daily Financial Data...")
time.sleep(1.5)
print("⚙️ [SYSTEM STATUS] Executing Analytics Algorithm...\n")
time.sleep(1)

# 2. Advanced Financial Calculations
net_profit = total_sales - total_expense
profit_margin_percentage = (net_profit / total_sales) * 100

# Generating the Final Professional Business Report
print("-" * 55)
print(f"💰 TOTAL REVENUE GENERATED    :  $ {total_sales:,.2f} USD")
print(f"📉 TOTAL EXPENSES INCURRED    :  $ {total_expense:,.2f} USD")
print("-" * 55)
print(f"🏆 NET PROFIT EARNED          :  $ {net_profit:,.2f} USD")
print(f"📈 NET PROFIT MARGIN RATIO    :  {profit_margin_percentage:.2f} %")
print("-" * 55 + "\n")

# 3. AI Voice Assistant Integration
# The AI engine will read out the final financial report automatically
voice_narration_text = (
    f"Congratulations Mortoja! Today's total net profit is {int(net_profit)} dollars. "
    f"Your business net profit margin ratio is {profit_margin_percentage:.1f} percent. "
    f"The automated financial system report is successfully generated."
)

print("🔊 [AI ENGINE] Generating Voice Narrative Report...")
tts = gTTS(voice_narration_text, lang='en')
tts.save("profit_analytics_report.mp3")

print("✨ [SUCCESS] Enterprise Financial Report and AI Audio are Ready Below:")
print("=" * 55)
ipd.Audio("profit_analytics_report.mp3")
```
